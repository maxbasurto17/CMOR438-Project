## K-Means Clustering Example: Clustering WR Archetypes ##

The wide receiver (WR) is one of the most versatile positions in football, both in terms of size and role within a team. Some receivers are used only in the short passing game, while others are used primarily in the redzone or deep balls. In this example, we seek to use K-Means clustering to identify different prototypical groups of WRs, and how these prototypes are used on the field.   

```python
import numpy as np
import rice_ml
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
```

First, we load in our game data, which contains receiving data from the NFL 2025-2026 season for all WRs who caught any passes. We will also use size data from our players dataset to inform us on how a player's size impacts his usage.  

```python
# Load in NFL receiving stats dataset
csv_path = Path("..") / "data" / "NFLReceivingStats.csv"
receiving = pd.read_csv(csv_path)
# Display the first 5 rows of the dataset
receiving.head(5)
```

```python
# Load in NFL player stats dataset
csv_path = Path("..") / "data" / "NFL_players.csv"
players = pd.read_csv(csv_path)

# Merge two datasets on pff_id = Player ID
merged = pd.merge(receiving, players, left_on="Player Id", right_on="pff_id")
merged.head(5)
```

Once we have merged the two datasets on the PFF ID as a primary key, we are ready to filter and clean the set. First we filter our data to only WRs. To ensure WR data is not skewed by a small sample of targets, we will remove all receivers with fewer than 15 targets in the season. Then, we select only the relevant ID and feature columns (Name, Player Id, POS, height, weight, YPR, YAC/REC, ADOT, TGT, CTT). In general, these feature columns are per target stats that are independent of volume. This way, WR classification will be impacted only by how a receiver is used when he is on the field. We will be doing some feature engineering with targets (TGT) and Contested Targets (CTT) to see what proportion of a WR's targets come in contested/jump ball situations. We will use that proportion for the model rather than the raw targets and contested targets to avoid volume stats. 

Our final features are as follows:

- Height: Player height in inches

- Weight: Player weight in pounds

- YPR: Yards per reception

- YAC/REC: Yards after Catch per reception

- ADOT: Average depth of target

- CTR: Contested Target Rate

```python
# Filter for wide receivers only
merged = merged[merged["POS"] == "WR"]
# Filter for players with at least 15 targets
merged = merged[merged["TGT"] >= 15]
# Select relevant features for clustering
features = merged[["Name", "Player Id", "POS", "height", "weight", "YPR", "YAC/REC", "ADOT", "TGT", "CTT"]]
# engineer Coontested target rate (CTR)
features["CTR"] = features["CTT"] / features["TGT"]
# Drop targets and CTT
features = features.drop(columns=["TGT", "CTT"])

features.head(5)
```

Once we have developed and selected our features, we are ready to standardize our data. We will be using the rice_ml built in Standard Scaler to standardize our features and save the mean and standard deviation of each feature. Once we have finished clustering on the standardized data, we can reconstruct our cluster means for each feature. 

```python
scaler = rice_ml.StandardScaler()
scaled_features = scaler.fit_transform(features[["height", "weight", "YPR", "YAC/REC", "ADOT", "CTR"]]).to_numpy()
# print first 5 rows of scaled features
print(scaled_features[:5])
```

Now, we are ready to fit our KMeans model. To initialize the KMeans class, we need to provide arguements for the number of clusters (k) and the maximum number of iterations through our dataset. 

We are hypothesizing that we will have three natural types of WRs (short-area/gadget WRs, intermediate route-runners, and deep threats). As a result, we are choosing k = 3. We will use the default max iterations of 100 for now. We also set the seed to 42 outside of the model for result reproducability.

To fit our model, we simply call the fit method with our standardized data as an arguement. This will automatically create three clusters using our KMeans algorithm and save the results. It is important to note that the algorithm uses euclidean distance to fit the model and label points. This is why it is imperative that we use standardized features to avoid feature scale or magnitude issues when labeling points.

```python
# Set seed to 42
np.random.seed(42)

# Initialize KMeans with 3 clusters and a maximum of 100 iterations
kmeans = rice_ml.KMeans(k=3, max_iter=100)

# Fit KMeans to the scaled features
kmeans.fit(scaled_features)
```

To inspect our model performance, we can extract the cluster centers and convert them back to real-world values.

```python
# Extract the scaled cluster centers
scaled_centers = kmeans.centers_

# Add .to_numpy() to the scaler attributes to prevent Pandas broadcasting errors
original_centers = (scaled_centers * scaler.std_.to_numpy()) + scaler.mean_.to_numpy()

# Put it in a DataFrame
features_list = ["height", "weight", "YPR", "YAC/REC", "ADOT", "CTR"]
centers_df = pd.DataFrame(original_centers, columns=features_list)

print("Original Cluster Centers:")
print(centers_df)
```

While these results reflect some of our pre-existing beliefs, they do not support all of them. It appears that group 0 is primarily shorter WRs who catch a lot of short passes and get lots of yards after the catch. This correlates strongly with our pre-determined Slot category. The other two categories are much harder to define at a glance, since they don't fit the typical statistical assumptions about route-runners or deep threats. 

To see if another breakdown works better, we will run the model for k = 4 to see if our results are more intuitive.

```python
np.random.seed(42)
# Initialize KMeans with 4 clusters and a maximum of 1000 iterations
kmeans_4 = rice_ml.KMeans(k=4, max_iter=100)

# Fit KMeans to the scaled features
kmeans_4.fit(scaled_features)

# Extract the scaled cluster centers
scaled_centers_4 = kmeans_4.centers_

# Add .to_numpy() to the scaler attributes to prevent Pandas broadcasting errors
original_centers_4 = (scaled_centers_4 * scaler.std_.to_numpy()) + scaler.mean_.to_numpy()

# Put it in a DataFrame
features_list = ["height", "weight", "YPR", "YAC/REC", "ADOT", "CTR"]
centers_df_4 = pd.DataFrame(original_centers_4, columns=features_list)

print("Original Cluster Centers:")
print(centers_df_4)
```

These results seem a lot more interpretable. 

Cluster 0 seems to be the typical Slot: low target depth, high YAC, and typically shorter and smaller WRs. 

Cluster 1 is likely the typical speedster/deep threat: Shorter and smaller than average, but high speed and a lot of deep targets. These players are asked to outrun their defender rather than outjump them. 

Cluster 2 is your prototypical outside route-runner. These are bigger outside WRs who get a lot of short and intermediate depth targets. They are asked to run sharp routes and create space between themselves and the defenders. 

Cluster 3 are the prototypical jump-ball threats. They are much bigger than the average receiver and are asked to win deep throws and contested jump balls. 

We can check these clusters by labeling the WRs in our dataset and seeing how our knowledge of a player's playstyle compares to his model defined category.

```python
# Label each player with their assigned cluster
features["Cluster"] = kmeans.labels_
# Display the features DataFrame with cluster labels
features
```

Looking at these results, the model is "correctly" identifying some players' playstyle, however we would personally disagree with many of these classifications. More specifically,  players in category 0 seem to be generally accurately labeled. Many of these players are true Slots and perfectly fit our label interpretation. Some of the category 0 WRs are Zay Flowers, Amon-Ra St. Brown, Hunter Renfrow, and Wan'Dale Robinson. There are, however, some WRs in this category that would typically be described as outside route runners. For instance, Justin Jefferson, Ja'Marr Chase, and Garrett Wilson are all classified in this category. This "mislabeling" extends to every category, where obvious choices are classified correctly, but some WRs are in unexpected clusters. 

It is possible that these clusters are accurate, and certain WRs' roles are misrepresented by popular opinion. But to us, it seems that there are errors in the model itself, possibly linked to a high-dimension feature space and improper weighting of features resulting from a universal distance function.
